10 more loc questions. Basic to Medium. Each question has its own DataFrame.

## Question 1: Two conditions

### Dataset
```python
df=pd.DataFrame({"Name":["Ali","Sara","John","David","Mary"],"Department":["Sales","HR","Sales","IT","HR"],"Salary":[50000,45000,70000,60000,55000]})
```

### Task
Using loc, select Sales employees with salary >= 60000.

````python
import pandas as pd
df=pd.DataFrame({"Name":["Ali","Sara","John","David","Mary"],"Department":["Sales","HR","Sales","IT","HR"],"Salary":[50000,45000,70000,60000,55000]})
df.loc[(df['Department']=='Sales')&
        (df['Salary']>= 60000)]
````

## Question 5: Update condition

Using loc, increase salary by 5000 for Sales employees.

## Question 6: Create column

### Dataset
```python
df=pd.DataFrame({"Name":["Ali","Sara","John","David"],"Salary":[50000,45000,70000,60000]})
```

### Task
Using loc, create Category: High if salary >= 60000, otherwise Low.

````python
df=pd.DataFrame({"Name":["Ali","Sara","John","David"],"Salary":[50000,45000,70000,60000]})
df.loc[df["Salary"]>= 60000, 'Category'] = 'High'
df.loc[df["Salary"] < 60000, 'Category'] = 'Low'
df
````

## Question 8: Missing values

df=pd.DataFrame({"Name":["Ali","Sara","John","David"],"Department":["Sales","HR","Sales","IT"],"Manager":["Ravi",None,"Amit",None]})

Using loc, replace missing Manager values with Not Assigned.

## Question 10: Interview filtering

### Dataset
```python
df=pd.DataFrame({"Name":["Ali","Sara","John","David","Mary","Aisha"],"Department":["Sales","HR","Sales","IT","Finance","Sales"],"Salary":[50000,45000,70000,60000,55000,65000],"Experience":[2,1,4,3,2,5]})
```

### Task
Using loc, return Name, Salary, Experience for Sales employees with Experience >= 3 and Salary > 55000.

````python
df=pd.DataFrame({"Name":["Ali","Sara","John","David","Mary","Aisha"],"Department":["Sales","HR","Sales","IT","Finance","Sales"],"Salary":[50000,45000,70000,60000,55000,65000],"Experience":[2,1,4,3,2,5]})
df.loc[(df['Department']== "Sales")&
        (df['Experience']>= 3) &
        (df['Salary']>55000) ,[ 'Name','Salary','Experience']]
````
