## Question 10: Filter + Update

### Dataset
```python
df=pd.DataFrame({"Name":["Ali","Sara","John","David"],"Department":["Sales","HR","Sales","IT"],"Salary":[50000,45000,70000,60000]})
```

### Task
Using loc, change Department to Senior Sales for employees whose salary is greater than 60000.

````python
df=pd.DataFrame({"Name":["Ali","Sara","John","David"],"Department":["Sales","HR","Sales","IT"],"Salary":[50000,45000,70000,60000]})
df.loc[(df['Department']=='Sales')&
        (df['Salary']> 60000),['Department']]='Seniour Sales'
````
