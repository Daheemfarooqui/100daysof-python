## Q13 — Department total

**Calculate total salary for each department using a loop.**

```python
employees={'Ali':{'salary':50000,'department':'IT'},'Sara':{'salary':65000,'department':'Sales'},'John':{'salary':55000,'department':'IT'},'Neha':{'salary':70000,'department':'Sales'},'Rahul':{'salary':62000,'department':'HR'}}
```

```python
dept_salary = {}

for name, details in employees.items():
    department = details['department']
    salary = details['salary']

    if department not in dept_salary:
        dept_salary[department] = 0

    dept_salary[department] += salary

print(dept_salary)
```

## Q20 — Interview challenge

**Calculate average salary of Sales employees with salary >= 60000. Use loops and nested dictionaries. Do not use `groupby()`, `sum()`, or `mean()`.**

```python
employees={'Ali':{'salary':50000,'department':'IT'},'Sara':{'salary':65000,'department':'Sales'},'John':{'salary':55000,'department':'IT'},'Neha':{'salary':70000,'department':'Sales'},'Rahul':{'salary':62000,'department':'Sales'}}
```

```python
count = 0
sum_salary = 0
for name , details in employees.items():
    salary = details['salary']
    dept   = details['department'] 
    if dept == 'Sales' and salary >= 60000:
        sum_salary += salary
        count += 1
avg_salary = sum_salary/ count

print(avg_salary)
```
