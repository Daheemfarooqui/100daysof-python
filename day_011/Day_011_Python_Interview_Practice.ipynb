## Question 9 (Functions)

### Dataset
```python
employees={"Ali":50000,"Sara":45000,"John":70000}
```

### Task
Write high_salary(data) to return employee names with salary > 50000.

````python
employees={"Ali":50000,"Sara":45000,"John":70000}
def high_salary(employees):
    new_list = [i for i , salary in employees.items() if salary > 50000]
    return new_list
high_salary(employees)
````
