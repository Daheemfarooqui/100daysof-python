## Question 19 — DataFrame Loop + List

## Question 20 — Interview Challenge

**Task:** Using a loop, calculate average salary of Sales employees. Do not use groupby(), mean(), or sum().

```python
import pandas as pd

df = pd.DataFrame({'Name':['Ali','Sara','John','David','Mary'],'Department':['Sales','HR','Sales','IT','Sales'],'Salary':[50000,45000,70000,60000,55000]})
```

```python
sum_value = 0
count = 0
avg_value = 0
for i in df[df['Department']=='Sales']['Salary']:
    sum_value += i
    count += 1
    avg_value =sum_value/count 
print(avg_value)
    
    
```
