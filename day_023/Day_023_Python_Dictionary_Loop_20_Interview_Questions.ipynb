## Q14 — Product revenue

## Q20 — Final interview

**Find average salary of Sales employees whose salary is above the overall company average. Do not use groupby(), sum(), mean(), max(), or sorted().**

```python
e={'Ali':{'salary':50000,'dept':'IT'},'Sara':{'salary':65000,'dept':'Sales'},'John':{'salary':72000,'dept':'IT'},'Neha':{'salary':80000,'dept':'Sales'},'Rahul':{'salary':62000,'dept':'HR'},'Aman':{'salary':75000,'dept':'Sales'},'Zoya':{'salary':55000,'dept':'HR'}}
```

```python

total_salary = 0
count = 0

for name, details in e.items():
    total_salary += details['salary']
    count += 1

company_avg = total_salary / count

sales_total = 0
sales_count = 0

for name, details in e.items():
    salary = details['salary']
    dept = details['dept']

    if dept == 'Sales' and salary > company_avg:
        sales_total += salary
        sales_count += 1

avg_salary = sales_total / sales_count

print(avg_salary)
```
