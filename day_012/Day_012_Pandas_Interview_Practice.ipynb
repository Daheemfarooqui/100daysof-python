## Question 1: groupby() - Total Revenue

### Dataset
```python
df = pd.DataFrame({
    "Department":["Sales","HR","Sales","IT","HR"],
    "Revenue":[50000,40000,70000,60000,45000]
})
```

### Task
Using groupby(), find total Revenue for each Department.

````python
import pandas as pd
df = pd.DataFrame({
    "Department":["Sales","HR","Sales","IT","HR"],
    "Revenue":[50000,40000,70000,60000,45000]
})
df.groupby(by='Department')['Revenue'].sum()


````

## Question 2: groupby() - Average Salary

### Dataset
```python
df = pd.DataFrame({
    "Department":["Sales","Sales","HR","HR","IT"],
    "Salary":[50000,70000,45000,55000,60000]
})
```

### Task
Using groupby(), find average Salary for each Department.

````python
df = pd.DataFrame({
    "Department":["Sales","Sales","HR","HR","IT"],
    "Salary":[50000,70000,45000,55000,60000]
})

df.groupby(by ="Department")['Salary'].mean()
````

## Question 3: agg() - Multiple Metrics

### Dataset
```python
df = pd.DataFrame({
    "Department":["Sales","Sales","HR","HR","IT"],
    "Salary":[50000,70000,45000,55000,60000]
})
```

### Task
For each Department, find minimum, maximum and average Salary using agg().

````python

df = pd.DataFrame({
    "Department": ["Sales", "Sales", "HR", "HR", "IT"],
    "Salary": [50000, 70000, 45000, 55000, 60000]
})

result = df.groupby("Department")["Salary"].agg(["min", "max", "mean"])
result
````

## Question 6: loc

### Dataset
```python
df = pd.DataFrame({
    "Name":["Ali","Sara","John","David"],
    "Department":["Sales","HR","Sales","IT"],
    "Salary":[50000,45000,70000,60000]
})
```

### Task
Using loc, select Name and Salary of Sales employees whose salary is greater than 55000.

````python

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "John", "David"],
    "Department": ["Sales", "HR", "Sales", "IT"],
    "Salary": [50000, 45000, 70000, 60000]
})

result = df.loc[
    (df["Department"] == "Sales") & (df["Salary"] > 55000),
    ["Name", "Salary"]
]

print(result)
````

## Question 8: merge()

### Dataset
```python
employees = pd.DataFrame({
    "emp_id":[1,2,3],
    "name":["Ali","Sara","John"]
})
salary = pd.DataFrame({
    "emp_id":[1,2,3],
    "salary":[50000,45000,70000]
})
```

### Task
Merge employees and salary using emp_id.

````python
employees = pd.DataFrame({
    "emp_id":[1,2,3],
    "name":["Ali","Sara","John"]
})
salary = pd.DataFrame({
    "emp_id":[1,2,3],
    "salary":[50000,45000,70000]
})
employees.merge(salary,how = "inner",on='emp_id')


````

## Question 10: apply()

### Dataset
```python
df = pd.DataFrame({
    "Name":["Ali","Sara","John"],
    "Salary":[50000,60000,70000]
})
```

### Task
Create a new bonus column where bonus is 10% of Salary using apply().

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "John"],
    "Salary": [50000, 60000, 70000]
})

df["Bonus"] = df["Salary"].apply(lambda x: x * 0.10)

print(df)
````
