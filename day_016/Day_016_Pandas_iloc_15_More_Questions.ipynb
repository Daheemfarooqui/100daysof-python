## Question 1

Select rows 2, 4 and 5.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ['Ali', 'Sara', 'John', 'David', 'Mary', 'Aisha'],
    "Department": ['Sales', 'HR', 'IT', 'Sales', 'Finance', 'HR'],
    "Salary": [50000, 45000, 70000, 60000, 55000, 48000]
})
```
