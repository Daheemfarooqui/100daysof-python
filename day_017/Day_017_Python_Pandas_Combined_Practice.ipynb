## Question 1 — For Loop

**Task:** Print employee names whose salary is greater than 60000.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'John', 'David'],
    'Salary': [50000, 45000, 70000, 60000]
})
```

```python
for _,i in df.iterrows():
    if i['Salary'] > 60000:
        print(i['Name'])
```

## Question 3 — Function

**Task:** Write average_salary(df) using a loop and return the average salary.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'John'],
    'Salary': [50000, 60000, 70000]
})
```

```python
def average_salary(df):
    count = 0
    sum_ = 0
    for i in df['Salary']:
        count += 1
        sum_ += i
    avg_ = sum_/count
    return avg_
average_salary(df)

        
```

## Question 15 — Combined Interview

**Task:** Write process_data(df) that uses a loop to create a list of employee names with salary >= 60000. Return the list.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'John', 'David', 'Mary'],
    'Department': ['Sales', 'HR', 'Sales', 'IT', 'Finance'],
    'Salary': [50000, 45000, 70000, 60000, 55000]
})
```

```python
def process_data(df):
    new_list = []
    for _,i in df.iterrows():
        if i["Salary"] >= 60000:
            new_list.append([i['Name'],i['Salary']])
    return new_list

process_data(df)    
            
```
