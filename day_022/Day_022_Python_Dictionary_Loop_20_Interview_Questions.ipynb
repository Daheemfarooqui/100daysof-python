## Q8 — Department average

**Using nested dictionaries and loops, calculate average salary for each department.**

```python
e={'Ali':{'salary':50000,'dept':'IT'},'Sara':{'salary':65000,'dept':'Sales'},'John':{'salary':55000,'dept':'IT'},'Neha':{'salary':70000,'dept':'Sales'},'Rahul':{'salary':62000,'dept':'HR'}}
```

```python

avg_salary = {}
sum_salary = {}
count = {}
for name, details in e.items():
    salary = details['salary']
    dept = details['dept']

    if dept not in sum_salary:
        sum_salary[dept] = 0
        count[dept] = 0

    
    sum_salary[dept] += salary
    count[dept] += 1
    avgsalary = sum_salary[dept]/count[dept]
    avg_salary[dept] = avgsalary

print(avg_salary)
    
```

## Q13 — Monthly average

## Q19 — Performance level

**Calculate total sales and mark Excellent >=60000, Good >=40000, otherwise Needs Improvement.**

```python
e={'Ali':{'sales':[15000,18000,20000]},'Sara':{'sales':[22000,21000,23000]},'John':{'sales':[10000,12000,11000]},'Neha':{'sales':[25000,24000,22000]}}
```

```python
sum_sales = {}

for name, details in e.items():
    sale = details['sales']
    total = sum(sale)

    if total >= 60000:
        rating = 'excellent'
    elif total >= 40000:
        rating = 'good'
    else:
        rating = 'needs improvement'

    sum_sales[name] = {
        'total_sales': total,
        'rating': rating
    }

print(sum_sales)
```
