## Question 20 — Combined Interview

**Task:** Write process_sales(df) that filters Sales employees with salary >= 55000 and returns their names as a list. Use a function and loc/filtering.

```python
import pandas as pd

df=pd.DataFrame({"Name":["Ali","Sara","John","David","Mary"],"Department":["Sales","HR","Sales","IT","Sales"],"Salary":[50000,45000,70000,60000,55000]})
```

```python
def process_sales(df):
    names_emp = df.loc[
        (df['Department']=='Sales')&
        (df['Salary']>=55000)]['Name'].to_list()
    return names_emp
process_sales(df)
```
